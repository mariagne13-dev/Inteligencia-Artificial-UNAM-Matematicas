Este código utiliza el algoritmo de poda alfa-beta en el juego del tic-tac-toe. Puedes probarlo tal como está.



https://www.youtube.com/watch?v=I0y-TGehf-4

```python
import math

def imprimir_tablero(tablero):
    """Muestra el tablero de forma legible."""
    for fila in tablero:
        print(" | ".join(fila))
        print("-" * 9)
    print()

def es_tablero_lleno(tablero):
    """Regresa True si ya no hay casillas vacías."""
    return all(" " not in fila for fila in tablero)

def evaluar_estado(tablero):
    """
    Evalúa el estado del juego:
      1  -> gana X
     -1  -> gana O
      0  -> empate
      None -> el juego continúa
    """
    lineas = []

    # Filas
    lineas.extend(tablero)

    # Columnas
    for col in range(3):
        lineas.append([tablero[0][col], tablero[1][col], tablero[2][col]])

    # Diagonales
    lineas.append([tablero[0][0], tablero[1][1], tablero[2][2]])
    lineas.append([tablero[0][2], tablero[1][1], tablero[2][0]])

    for linea in lineas:
        if linea == ["X", "X", "X"]:
            return 1
        if linea == ["O", "O", "O"]:
            return -1

    if es_tablero_lleno(tablero):
        return 0

    return None

def alphabeta(tablero, profundidad, alfa, beta, es_maximizando):
    """
    Algoritmo Minimax con poda Alpha-Beta.

    X es el jugador MAX: intenta maximizar el valor.
    O es el jugador MIN: intenta minimizar el valor.
    """
    estado_actual = evaluar_estado(tablero)

    if estado_actual is not None:
        return estado_actual

    if es_maximizando:
        mejor_valor = -math.inf

        for i in range(3):
            for j in range(3):
                if tablero[i][j] == " ":
                    tablero[i][j] = "X"
                    valor = alphabeta(
                        tablero, profundidad + 1, alfa, beta, False
                    )
                    tablero[i][j] = " "

                    mejor_valor = max(mejor_valor, valor)
                    alfa = max(alfa, mejor_valor)

                    # Poda beta
                    if beta <= alfa:
                        return mejor_valor

        return mejor_valor

    else:
        mejor_valor = math.inf

        for i in range(3):
            for j in range(3):
                if tablero[i][j] == " ":
                    tablero[i][j] = "O"
                    valor = alphabeta(
                        tablero, profundidad + 1, alfa, beta, True
                    )
                    tablero[i][j] = " "

                    mejor_valor = min(mejor_valor, valor)
                    beta = min(beta, mejor_valor)

                    # Poda alfa
                    if beta <= alfa:
                        return mejor_valor

        return mejor_valor

def mejor_movimiento_ia(tablero):
    """
    Devuelve la mejor jugada para O.

    Como O es MIN, buscamos el movimiento con el MENOR valor.
    """
    mejor_valor = math.inf
    mejor_mov = None
    alfa = -math.inf
    beta = math.inf

    for i in range(3):
        for j in range(3):
            if tablero[i][j] == " ":
                tablero[i][j] = "O"

                # Después de O juega X, que maximiza.
                valor = alphabeta(tablero, 0, alfa, beta, True)

                tablero[i][j] = " "

                if valor < mejor_valor:
                    mejor_valor = valor
                    mejor_mov = (i, j)

                beta = min(beta, mejor_valor)

    return mejor_mov

# Juego: Humano = X, IA = O
tablero_inicial = [
    [" ", " ", " "],
    [" ", " ", " "],
    [" ", " ", " "]
]

turno = "X"

while evaluar_estado(tablero_inicial) is None:
    imprimir_tablero(tablero_inicial)

    if turno == "X":
        try:
            fila, col = map(
                int,
                input("Turno de X. Ingresa fila y columna (0, 1 o 2): ").split()
            )

            if fila not in range(3) or col not in range(3):
                print("Posición fuera del tablero. Intenta de nuevo.")
                continue

            if tablero_inicial[fila][col] != " ":
                print("Casilla ocupada. Intenta de nuevo.")
                continue

            tablero_inicial[fila][col] = "X"

        except ValueError:
            print("Ingresa dos números separados por espacio.")
            continue

    else:
        print("Turno de O (IA)")
        movimiento = mejor_movimiento_ia(tablero_inicial)

        if movimiento is not None:
            fila, col = movimiento
            tablero_inicial[fila][col] = "O"

    turno = "O" if turno == "X" else "X"

imprimir_tablero(tablero_inicial)

resultado = evaluar_estado(tablero_inicial)

if resultado == 1:
    print("¡X gana!")
elif resultado == -1:
    print("¡O gana!")
else:
    print("¡Empate!")
```

## ¿Qué hace la poda Alpha-Beta?

La poda **Alpha-Beta** es una optimización del algoritmo **Minimax**.  
Minimax analiza posibles jugadas futuras suponiendo que:

- **MAX** intenta obtener el valor más grande posible.
- **MIN** intenta obtener el valor más pequeño posible.

En este ejercicio:

- `X` es **MAX**, por lo que una victoria de `X` vale `1`.
- `O` es **MIN**, por lo que una victoria de `O` vale `-1`.
- Un empate vale `0`.

Los parámetros principales son:

- **alpha (`alfa`)**: mejor valor que MAX puede garantizar hasta el momento.
- **beta (`beta`)**: mejor valor que MIN puede garantizar hasta el momento.

Cuando se cumple

\[
\beta \leq \alpha,
\]

ya no es necesario explorar las ramas restantes del árbol porque no pueden cambiar la decisión óptima.  
A esto se le llama **poda**.

La ventaja es que se obtiene la misma decisión que con Minimax, pero explorando menos estados del juego.
